## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data
Pada tahap ini, dilakukan pembersihan data berdasarkan hasil pemeriksaan pada tahap Assessing Data. Proses ini meliputi penyesuaian tipe data serta penanganan data anomali (*outlier)*

#### Fixing df_day problem
**Memperbaiki Tipe Data** : `dteday` memiliki tipe data `str`, padahal berisi informasi tanggal. Oleh karena itu, tipe datanya perlu diubah menjadi `datetime` agar dapat dikenali sebagai data tanggal dan memudahkan proses analisis berdasarkan waktu.

**Menangani Outliers (Anomali)** :
* `casual`: Berdasarkan pemeriksaan menggunakan metode Interquartile Range (IQR), ditemukan 44 nilai `casual` yang melebihi batas atas sebesar 2.266,75. Penanganan dilakukan menggunakan teknik Imputation dengan Boundary Value, yaitu mengganti nilai yang melebihi batas tersebut dengan nilai batas atas IQR.

* `Kolom cnt`: Setelah penanganan outlier pada `casual`, nilai cnt dihitung kembali berdasarkan penjumlahan `casual` dan `registered` untuk menjaga konsistensi data.


```python
# Memperbaiki Tipe Data
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
```

```python
#Menangani Outlier

# Menghitung batas atas menggunakan metode IQR untuk mengatasi outlier pada jumlah penyewaan casual (> 2.266,75)
Q1 = df_day['casual'].quantile(0.25)
Q3 = df_day['casual'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + (1.5 * IQR)

# Mengganti nilai outlier yang melewati batas atas dengan batas atas
df_day['casual'] = df_day['casual'].mask(
    df_day['casual'] > upper_bound,
    upper_bound
)

# Memperbarui total penyewaan
df_day['cnt'] = df_day['casual'] + df_day['registered']
```

```python
# Mengelompokkan data berdasarkan musim dan tahun untuk menghitung rata-rata penyewaan
seasonal_summary = df_day.groupby(
    by=['season', 'yr']
)['cnt'].mean().unstack()

seasonal_summary.columns = ['2011', '2012']

seasonal_summary['growth_percentage'] = (
    (seasonal_summary['2012'] - seasonal_summary['2011'])
    / seasonal_summary['2011']
) * 100

print("--- Rata-Rata Penyewaan Sepeda Berdasarkan Musim ---")

for season, row in seasonal_summary.iterrows():
    print(f"\nMusim {season}:")
    print(f"  Rata-rata penyewaan 2011: {row['2011']:.2f}")
    print(f"  Rata-rata penyewaan 2012: {row['2012']:.2f}")
    print(f"  Pertumbuhan: {row['growth_percentage']:.2f}%")
```

Untuk mengetahui perkembangan penyewaan sepeda berdasarkan musim, dilakukan perbandingan rata-rata jumlah **penyewaan harian antara tahun 2011 dan 2012**. Perbandingan ini bertujuan untuk mengidentifikasi musim dengan persentase pertumbuhan penyewaan tertinggi serta melihat perubahan tingkat penyewaan pada setiap musim.

Berdasarkan hasil perhitungan, diperoleh beberapa temuan pada setiap musim:

- Musim Gugur `(season = 3)`: Mencatat rata-rata penyewaan tertinggi pada kedua tahun, yaitu 4.451,39 pada tahun 2011 dan 6.783,32 pada tahun 2012. Persentase pertumbuhannya mencapai 52,39%.

- Musim Dingin `(season = 4)`: Rata-rata penyewaan meningkat dari 3.663,00 menjadi 5.767,01, dengan pertumbuhan sebesar 57,44%.

**Insight:**
- Musim semi mencatat pertumbuhan penyewaan tertinggi, yaitu sebesar 111,29%. Artinya, rata-rata penyewaan pada musim tersebut meningkat lebih dari dua kali lipat dibandingkan tahun sebelumnya.

Insight:

Sementara itu, kondisi hujan atau salju ringan hanya mencakup sekitar 2,87%. Hal ini menunjukkan bahwa distribusi data berdasarkan kondisi cuaca tidak merata.
```python
# Mengelompokkan data berdasarkan kondisi cuaca dan menghitung statistik penyewaan
weather_summary = df_day.groupby(
    by='weathersit'
)['cnt'].agg(['mean', 'median', 'std', 'count'])

print("--- Statistik Penyewaan Berdasarkan Kondisi Cuaca ---")

for weather, row in weather_summary.iterrows():
    print(f"\nKondisi cuaca {weather}:")
    print(f"  Rata-rata penyewaan: {row['mean']:.2f}")
    print(f"  Median penyewaan: {row['median']:.2f}")
    print(f"  Standar deviasi: {row['std']:.2f}")
    print(f"  Jumlah hari pengamatan: {int(row['count'])}")
```

Untuk memahami perbedaan pola penyewaan sepeda pada setiap kondisi cuaca, dilakukan perbandingan statistik jumlah penyewaan harian berdasarkan kategori `weathersit`. ## Analisis Lanjutan
Analisis lanjutan ini menggunakan teknik Binning untuk mengelompokkan data berdasarkan karakteristik tertentu ke dalam beberapa kategori. Pengelompokan ini bertujuan untuk memperoleh informasi tambahan mengenai pola dan distribusi data berdasarkan kategori yang telah ditentukan.

```python
# Menentukan batas setiap kategori permintaan
bins = [
    df_day['cnt'].min() - 1,
    df_day['cnt'].quantile(1/3),
    df_day['cnt'].quantile(2/3),
    df_day['cnt'].max()
]

labels = ['Rendah', 'Sedang', 'Tinggi']

# Mengelompokkan jumlah penyewaan
df_day['rental_demand'] = pd.cut(
    df_day['cnt'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Menghitung jumlah hari pada setiap kategori
demand_distribution = df_day['rental_demand'].value_counts().sort_index()

print("--- Distribusi Tingkat Permintaan Penyewaan ---")

for category, count in demand_distribution.items():
    print(f"Permintaan {category}: {count} hari")
```

**Insight:**
- Distribusi jumlah hari pada ketiga kategori permintaan relatif seimbang, yaitu **244** hari untuk kategori Rendah, **243** hari untuk kategori Sedang, dan **244** hari untuk kategori Tinggi.

- Pembagian ini menghasilkan jumlah data yang relatif seimbang karena menggunakan teknik Binning berdasarkan kuantil.

- Pengelompokan ini dapat digunakan untuk melihat lebih lanjut bagaimana tingkat permintaan penyewaan berbeda berdasarkan musim maupun kondisi cuaca.

```python
season_demand = pd.crosstab(
    df_day['season'],
    df_day['rental_demand'],
    normalize='index'
) * 100

season_demand.index = [
    'Musim Semi', 'Musim Panas', 'Musim Gugur', 'Musim Dingin'
]

print("--- Persentase Permintaan Berdasarkan Musim ---")
print(season_demand.round(2).to_string())
```

**Insight:**
- Musim Semi didominasi permintaan rendah sebesar **77,35%**, sedangkan permintaan tinggi hanya **4,42%.**

- Musim Panas memiliki persentase permintaan tinggi sebesar 42,39%, lebih besar dibandingkan permintaan rendah.

- Musim Gugur menunjukkan persentase permintaan tinggi terbesar, yaitu 50,53%, sehingga menjadi musim dengan kecenderungan permintaan penyewaan tertinggi.

- Musim Dingin memiliki distribusi permintaan yang lebih merata, dengan kategori tinggi sebesar 35,39%.

Permintaan penyewaan cenderung meningkat dari Musim Semi menuju Musim Gugur. Hal ini menunjukkan adanya pola perbedaan tingkat permintaan berdasarkan musim.

```python
weather_demand = pd.crosstab(
    df_day['weathersit'],
    df_day['rental_demand'],
    normalize='index'
) * 100

weather_demand.index = [
    'Cerah', 'Berkabut/Berawan', 'Hujan/Salju Ringan'
]

print("--- Persentase Permintaan Berdasarkan Cuaca ---")
print(weather_demand.round(2).to_string())
```
